### 2. Initialize the NER Pipeline
We use a pre-trained BERT model fine-tuned on the CoNLL-2003 dataset for NER (`dbmdz/bert-large-cased-finetuned-conll03-english`).

```python
from transformers import pipeline

# Initialize the pipeline
ner_pipeline = pipeline(
    'ner',
    model='dbmdz/bert-large-cased-finetuned-conll03-english',
    aggregation_strategy='simple'
)
```

### 3. Test NER on Sample Text

```python
sample_text = "Google was founded in September 1998 by Larry Page and Sergey Brin while they were Ph.D. students at Stanford University in California."

results = ner_pipeline(sample_text)

# Display the structured results
import pandas as pd
df_results = pd.DataFrame(results)
display(df_results)
```
