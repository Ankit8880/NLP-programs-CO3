## 4. Classification Model Strategy & Pipeline
We will construct a TF-IDF Vectorizer combined with a linear classifier (such as SGDClassifier or Logistic Regression) to categorize our documents.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Sample Synthetic Dataset
data = {
    'text': [
        'invoice payment due receipt total amount bill details tax account balance statement',
        'resume profile curriculum vitae skills python machine learning experience data scientist job education',
        'service contract agreement terms conditions signature party obligations lease client partner',
        'invoice payment tax statement receipt bill total billing invoice',
        'work experience skills education CV developer software engineering computer science job',
        'legal contract signed lease partnership NDA agreement client clauses'
    ],
    'label': ['invoice', 'resume', 'contract', 'invoice', 'resume', 'contract']
}

df = pd.DataFrame(data)

# Define the pipeline
model_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('clf', LogisticRegression())
])

# Train model on synthetic data
model_pipeline.fit(df['text'], df['label'])
print("Model training simulated successfully!")
```

## 5. Automatic Separation Logic
Using the trained model, we classify an input document and automatically route (separate) the file to designated folders.

```python
import shutil

def classify_and_separate_file(file_path, output_dir="separated_documents"):
    # 1. Extract and Clean text
    # raw_text = extract_text_from_pdf(file_path) # uncomment in production
    sample_raw_text = "This is a legal contract agreement signed with our client for software services."
    cleaned = clean_text(sample_raw_text)

    # 2. Predict Category
    predicted_category = model_pipeline.predict([cleaned])[0]
    print(f"Document predicted category: {predicted_category.upper()}")

    # 3. Create target directory and separate/move file
    target_dir = os.path.join(output_dir, predicted_category)
    os.makedirs(target_dir, exist_ok=True)

    # Simulate routing
    simulated_dest = os.path.join(target_dir, os.path.basename(file_path))
    print(f"Routing file '{file_path}' to -> '{simulated_dest}'")

# Run mock separation execution
classify_and_separate_file("sample_contract.pdf")
```

## 3. Document Extraction and Preprocessing
We will define functions to extract text content from PDFs directly, which will serve as our features for the classification ML model.

```python
def extract_text_from_pdf(pdf_path):
    """Extract text from a given PDF file."""
    text = ""
    try:
        reader = PdfReader(pdf_path)
        for page in reader.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"
    except Exception as e:
        print(f"Error reading {pdf_path}: {e}")
    return text.strip()

# Quick text cleaning function
def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```
